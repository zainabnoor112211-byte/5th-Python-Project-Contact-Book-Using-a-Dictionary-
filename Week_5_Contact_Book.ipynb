```python
## Complete Contact Book Program
```

```python
# Empty dictionary for contacts
contacts = {}

print("===== CONTACT BOOK =====")

while True:
    # Display menu
    print("\n1. Add Contact")
    print("2. Search Contact")
    print("3. Exit")

    choice = input("Enter your choice: ").strip()

    # Add Contact
    if choice == "1":
        name = input("Enter contact name: ").strip().title()
        phone = input("Enter phone number: ").strip()
        email = input("Enter email address: ").strip()

        # Check empty information
        if name == "" or phone == "" or email == "":
            print("All information is required.")

        else:
            contacts[name] = {
                "phone": phone,
                "email": email
            }

            print("Contact added successfully.")

    # Search Contact
    elif choice == "2":
        search_name = input(
            "Enter contact name to search: "
        ).strip().title()

        if search_name in contacts:
            print("\nContact found!")
            print("Name:", search_name)
            print("Phone:", contacts[search_name]["phone"])
            print("Email:", contacts[search_name]["email"])

        else:
            print("Contact not found.")

    # Exit
    elif choice == "3":
        print("Contact Book closed.")
        break

    # Invalid choice
    else:
        print("Invalid choice. Enter 1, 2, or 3.")
```
